```python
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
  return (celsius * 9/5) + 32
```

```python
# Ask the user for a temperature and a conversion threshold
temperature = float(input('Enter a temperature in Celsius: '))
threshold = float(input('Enter the threshold temperature in Celsius: '))
```

```python
# Convert the temperature
fahrenheit = celsius_to_fahrenheit(temperature)
print(f'{temperature} celsius is equal to {fahrenheit:.1f}fahrenheit.')
```

```python
# Compare the temperature with the threshold
if temperature > threshold:
  print('The temperature is above the threshold.')
if temperature < threshold:
      print('The temperature is below the threshold.')
else:
      print('The temperature is exactly at the threshold.')
```

```python
# Use a loop to convert several temperatures
print('\nConverting several temperatures: ')

for i in range(3):
  temp = float(input(f'Enter temperature {i + 1} in Celsius'))
  converted_temp = celsius_to_fahrenheit(temp)
  print(f'{temp} Celsius = {converted_temp: .1f}Fahrenheit')
```
